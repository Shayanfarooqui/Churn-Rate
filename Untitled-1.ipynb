```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

project_dir = r"C:\Users\muhammad.farooqui\Downloads\Python\Churn-Rate"
os.chdir(project_dir)


# Load all sheets from Excel file
excel_file = 'Customer_Churn_Data_Large - Lloyds.xlsx'
sheets = pd.read_excel(excel_file, sheet_name=None)

# Access individual dataframes
df1 = sheets['Customer_Demographics']
df2 = sheets['Transaction_History']
df3 = sheets['Customer_Service']
df4 = sheets['Online_Activity']
df5 = sheets['Churn_Status']
```

```python
#Sanity checks and normilization.
#Row and column counts and column names
for name, df in [
    ("df1_Customer_Demographics", df1),
    ("df2_Transaction_History", df2),
    ("df3_Customer_Service", df3),
    ("df4_Online_Activity", df4),
    ("df5_Churn_Status", df5)
]:
    print(f"\n==== {name} ====")
    print("Shape:", df.shape)                     # Number of rows & columns
    print("\nColumns:", df.columns.tolist())      # List of columns


# --- Basic display tuning for VS Code (optional) ---
pd.set_option('display.max_rows', 120)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 2000)

### Standardise column datatypes for dates
for d, cols in [
    (df2, ["TransactionDate"]),
    (df3, ["InteractionDate"]),
    (df4, ["LastLoginDate"]),
    (df5, ["ChurnDate"])  # if exists
]:
    for c in cols:
        if c in d.columns:
            d[c] = pd.to_datetime(d[c], errors="coerce")


# Categorical text -- Probably still dont need this step as data is clean already. Text → string dtype (robust categorical text)
for d in [df1, df3, df4]:
    obj_cols = d.select_dtypes(include="object").columns
    for c in obj_cols:
        d[c] = d[c].astype("string")



# Label as integer (0/1) if present
if "ChurnStatus" in df5.columns:
    df5["ChurnStatus"] = pd.to_numeric(df5["ChurnStatus"], errors="coerce").astype("Int64")


# --- (Optional but recommended) Deduplicate labels in df5 ---
# If multiple churn records per customer, keep the latest by ChurnDate; else keep last occurrence.
# if "CustomerID" in df5.columns:
#     if "ChurnDate" in df5.columns:
#         df5 = (df5.sort_values(["CustomerID", "ChurnDate"])
#                     .drop_duplicates(subset="CustomerID", keep="last"))
#     else:
#         df5 = df5.drop_duplicates(subset="CustomerID", keep="last")
```

```python
# Find duplicates based on 'CustomerID' in Customer_Demographics
dupes = df1.duplicated(subset=['CustomerID'], keep=False) 
print(df1[dupes].sort_values(by=['CustomerID']))

# Find duplicates based on 'CustomerID' in Churn_Status
dupes = df5.duplicated(subset=['CustomerID'], keep=False) 
print(df5[dupes].sort_values(by=['CustomerID']))

# df1 = sheets['Customer_Demographics']
# df2 = sheets['Transaction_History']
# df3 = sheets['Customer_Service']
# df4 = sheets['Online_Activity']
# df5 = sheets['Churn_Status']

# --- Duplicates by CustomerID (your checks, kept) ---
print("Duplicates in Customer Demograph by CustomerID:", df1.duplicated(subset="CustomerID").sum())
print("Duplicates in Transaction History by CustomerID:", df2.duplicated(subset="CustomerID").sum())
print("Duplicates in Customer Service by CustomerID:", df3.duplicated(subset="CustomerID").sum())
print("Duplicates in Online Activity by CustomerID:", df4.duplicated(subset="CustomerID").sum())
print("Duplicates in Churn Status by CustomerID:", df5.duplicated(subset="CustomerID").sum())
```

```python
# Left join to keep all customers; you can filter to labelled later.
df_demo = df1.merge(df5[["CustomerID", "ChurnStatus"]], on="CustomerID", how="left")

# --- Missingness summaries (DataFrames, not files) ---

def missing_summary_df(df: pd.DataFrame) -> pd.DataFrame:
    missing_count = df.isna().sum()
    return pd.DataFrame({"missing_count": missing_count}) \
             .sort_values("missing_count", ascending=False)


ms_df1 = missing_summary_df(df1)
ms_df2 = missing_summary_df(df2)
ms_df3 = missing_summary_df(df3)
ms_df4 = missing_summary_df(df4)
ms_df5 = missing_summary_df(df5)
ms_demo = missing_summary_df(df_demo)


print("\n missing (Customer_Demographics):\n", ms_df1.head(10))
print("\n missing (Transaction_History):\n", ms_df2.head(10))
print("\n missing (Customer_Service):\n", ms_df3.head(10))
print("\n missing (Online_Activity):\n", ms_df4.head(10))
print("\n missing (Churn_Status):\n", ms_df5.head(10))
print("\n missing (df_demo):\n", ms_demo.head(10))
```

```python
# --- Build your EDA base: Demographics + Churn ---

# --- EDA: quick stats ---
if "ChurnStatus" in df_demo.columns:
    print("\nChurnStatus counts:\n", df_demo["ChurnStatus"].value_counts(dropna=False))
    print("\nChurnStatus %:\n", (df_demo["ChurnStatus"].value_counts(normalize=True, dropna=False) * 100).round(2))

# --- Visuals (shown inline in VS Code Interactive) ---
sns.set_theme(style="whitegrid")

# 1) Class balance
if "ChurnStatus" in df_demo.columns:
    ax =sns.countplot(data=df_demo, x="ChurnStatus", hue="ChurnStatus", palette="Set2", legend=False)
    ax.set_title("Class Balance: ChurnStatus (0=No, 1=Yes)")
    plt.show()

# 2) Numeric distributions
num_cols = [c for c in df_demo.columns if pd.api.types.is_numeric_dtype(df_demo[c]) and c != "ChurnStatus"]
for col in num_cols[:6]:
    sns.histplot(df_demo[col].dropna(), kde=True)
    plt.title(f"Distribution: {col}")
    plt.show()

# 3) Churn rate by category (Gender, MaritalStatus, IncomeLevel)
for col in ["Gender", "MaritalStatus", "IncomeLevel"]:
    if col in df_demo.columns and "ChurnStatus" in df_demo.columns:
        tab = pd.crosstab(df_demo[col], df_demo["ChurnStatus"])
        tab["ChurnRate_%"] = (tab.get(1, 0) / tab.sum(axis=1) * 100).round(2)
        print(f"\n{col} — churn rate by category:\n", tab.sort_values("ChurnRate_%", ascending=False))
        ax = sns.barplot(x=tab.index.astype(str), y=tab["ChurnRate_%"].values, hue=tab.index.astype(str), palette="Reds", legend=False)
        ax.set_title(f"Churn rate by {col} (%)"); ax.set_ylabel("Churn rate (%)")
        plt.xticks(rotation=30, ha="right"); plt.show()

# 4) Boxplots: numeric vs churn
if "ChurnStatus" in df_demo.columns and num_cols:
    for col in num_cols[:6]:
        sns.boxplot(data=df_demo, x="ChurnStatus", y=col)
        plt.title(f"{col} by ChurnStatus")
        plt.show()

# 5) Correlation heatmap (numeric)
if len(num_cols) >= 2:
    corr = df_demo[num_cols + (["ChurnStatus"] if "ChurnStatus" in df_demo.columns else [])].corr(numeric_only=True)
    sns.heatmap(corr, cmap="vlag", center=0)
    plt.title("Correlation Heatmap (Numeric)")
    plt.show()
```

```python
##Agregated Customer Service interactions
agg_srv = df3.groupby("CustomerID").agg(
    total_interactions=("InteractionID", "count"),
    complaints=("InteractionType", lambda x: (x == "Complaint").sum()),
    unresolved=("ResolutionStatus", lambda x: (x == "Unresolved").sum())
).reset_index()

print("\nAggregated Customer Service interactions:\n", agg_srv.head(10))


##Agregated Online Activity
agg_online = df4.groupby("CustomerID").agg(
    login_frequency=("LoginFrequency", "mean"),
    last_login=("LastLoginDate", "max")
).reset_index()
print("\nAggregated Online Activity:\n", agg_online.head(10))

##Agregated Transaction History
agg_txn = df2.groupby("CustomerID").agg(
    txn_count=("TransactionID", "count"),
    total_amount_spent=("AmountSpent", "sum"),
    avg_amount_spent=("AmountSpent", "mean"),
    last_txn_date=("TransactionDate", "max")
).reset_index()
print("\nAggregated Transaction History:\n", agg_txn.head(10))
```

```python

df_demo = df1.merge(df5, on="CustomerID", how="left")
df_demo = df_demo.merge(agg_txn, on="CustomerID", how="left")

print("\nMerged Demographics with Churn and Transaction History:\n", df_demo.head(10))
```
